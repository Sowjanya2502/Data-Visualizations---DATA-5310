## Part 0: Adding Your Own Data to Choropleths in Altair

Part of the challenge of plotting data on a map is connecting geographic shapes (polygons made as a list of latitude/longitude points) to the data itself. Getting a list of the countries, states, or counties within the US, for example, is not hard. However, getting a data set containing all of their geographic borders, is. Altair/Vega has some geographic data built into it that is convenient to plot with. In the examples from the Altair demo, their geographic data is easily mapped to their informational data because it is nicely curated. 

**Problem: You download a data set that you want to plot on a map, using the built-in Altair geographies. You don't know how to connect the informational data to the geographic data.**

Note that the data has county and state name. The `counties` data links to a GeoJSON file, details can be found in the Altair notebook.

#### Looking at Geographic Data

We can look directly at the county-level geometries from the built-in Vega map. Notice that there are geometries and ID numbers. These ID numbers are FIPS (Federal Information Processing Standard) codes. Each state, county, and city has a unique FIPS code and it is used across federal governmental data. 

```python
countydf = gpd.read_file(
    data.us_10m.url, 
    driver='TopoJSON'
)
countydf.head()
```

#### Looking at Informational Data

There are many many sources of data that you might want to plot on geography. Some good starting points would be the US Census or other governmental organizations. This data was downloaded from the following link:
https://www.statista.com/statistics/241702/largest-counties-in-the-us/

```python
df = pd.read_excel('statistic_id241702_largest-counties-in-the-us-2022.xlsx', 
                   sheet_name = 'Data', 
                   header = 2, 
                   skiprows=[3,4], 
                   usecols = 'B:C'
                  )
df.rename(columns = {'Largest counties in the U.S. 2022':'County','Unnamed: 2':'Pop'}, inplace=True)
df[['County','State']] = df['County'].str.split(' County, ',expand=True)
df.tail()
```

#### Getting FIPS Codes

In order to merge these two data frames, we need to get the FIPS code for each county. We can do this with the `addfips` package. For another data type, for example, country names, you may have to do a merge by country names, which may be messy since there may be different spellings, abbreviations, missing data, etc. 

```python
df["FIPS"] = df.apply(lambda x: af.get_county_fips(x["County"], state=x["State"]), axis=1)
df.head()
```

#### Merging Data 
Now that the FIPS codes have been added, there are two ways we can merge these two data frames (geographic data and informational data). First, we *could* use a merge/join where we save the informational data into the geographic data frame. This would allow us to see the data all together and ensure we have merged correctly by cross-checking known values from each data frame. However, let's consider an example where we have various information we want to plot on a map (e.g. unemployment, poverty levels, education levels, and GDP). If we have many data sets to join, it may get prohibitively large and/or messy. The second way we can plot our new data is using a lookup transform. This temporarily looks up data from the informational frame to apply values based on the geographic (or any other!) data. Note here that the original data only has information for the top 100 most populous counties in the US. Notice that since our color encoding depends on the county, if there is no data available, the color will be null and therefore not plotted. 

```python
# counties = alt.topo_feature(data.us_10m.url, 'counties')

alt.Chart(countydf).mark_geoshape().encode(
    color="Pop:Q"
).transform_lookup(
    lookup='id',
    from_=alt.LookupData(df, 'FIPS', ["Pop"])
).project(
    type='albersUsa'
).properties(
    width=500,
    height=300
)
```

### Adding Data Points to a Folium Map

To put data onto our map, we can initialize markers to show location.  

```python
m = folium.Map(location=[47.6, -122.3], 
               tiles='cartodbpositron',
               zoom_start=12, control_scale=True)
for idx, row in df.iterrows():
    # Get lat and lon of points
    lon = row['X']
    lat = row['Y']

    # Add marker to the map
    folium.Marker(location=[lat, lon]).add_to(m)
m
```

### Adding Tooltips to a Folium Map

We can add more information in tooltip form easily by adding information in the marker. 

```python
m = folium.Map(location=[47.6, -122.3], 
               tiles='cartodbpositron',
               zoom_start=12, control_scale=True)
for idx, row in df.iterrows():
    # Get lat and lon of points
    lon = row['X']
    lat = row['Y']
    
    # get name for tooltip
    name = row['COMMON_NAME']

    # Add marker to the map
    folium.Marker(location=[lat, lon], 
                  tooltip=name
    ).add_to(m)
m
```

### Changing Icons in a Folium Map
We can change the appearance of the markers as a whole or conditionally as we add them. See the [marker icon options](https://github.com/lennardv2/Leaflet.awesome-markers) and find options within FontAwesome [here](https://fontawesome.com/search).

```python
m = folium.Map(location=[47.6, -122.3], 
               tiles='cartodbpositron',
               zoom_start=12, control_scale=True)

for idx, row in df.iterrows():
    # Get lat and lon of points
    lon = row['X']
    lat = row['Y']
    
    # get name for tooltip
    name = row['COMMON_NAME']

    # change icon based on tree height
    if row['TREEHEIGHT'] > 50:
        icon = folium.Icon(icon = 'mug-hot', color = 'beige', prefix = 'fa')
    else:
        icon = folium.Icon(icon = 'tree', color = 'darkgreen', prefix='fa')

    # Add marker to the map
    folium.Marker(location=[lat, lon], 
                  tooltip=name,
                  icon=icon
    ).add_to(m)
m
```

### Adding Filters to a Folium Map

We can add filters by using a `TagFilterButton` and giving criteria on which to filter. 

```python
m = folium.Map(location=[47.6, -122.3], 
               tiles='cartodbpositron',
               zoom_start=12, control_scale=True)


for idx, row in df.iterrows():
    # Get lat and lon of points
    lon = row['X']
    lat = row['Y']

    # get name for tooltip
    name = row['COMMON_NAME']

    # change icon based on tree height
    if row['TREEHEIGHT'] > 50:
        icon = folium.Icon(icon = 'person-walking', color = 'green', prefix = 'fa')
    else:
        icon = folium.Icon(icon = 'tree', color = 'darkgreen', prefix='fa')

    # get if the tree is heritage or not for filtering
    if row['HERITAGE']=='Y': 
        heritage_type = 'Heritage' 
    else: 
        heritage_type = 'Not Heritage'

    
    # Add marker to the map
    folium.Marker(location=[lat, lon], 
                  tooltip=name,
                  icon=icon,
                  tags=[heritage_type]
    ).add_to(m)


categories = ['Heritage','Not Heritage']
plugins.TagFilterButton(categories).add_to(m)

m
```

### Adding Point Clusters to a Folium Map
We can make the map cluster the points as we Zoom out by using `MarkerCluster`.

```python
m = folium.Map(location=[47.6, -122.3], 
               tiles='cartodbpositron',
               zoom_start=12, control_scale=True)

marker_cluster = plugins.MarkerCluster().add_to(m)

for idx, row in df.iterrows():
    # Get lat and lon of points
    lon = row['X']
    lat = row['Y']

    # get name for tooltip
    name = row['COMMON_NAME']

    # change icon based on tree height
    if row['TREEHEIGHT'] > 50:
        icon = folium.Icon(icon = 'person-walking', color = 'green', prefix = 'fa')
    else:
        icon = folium.Icon(icon = 'tree', color = 'darkgreen', prefix='fa')

    # get if the tree is heritage or not for filtering
    if row['HERITAGE']=='Y': 
        heritage_type = 'Heritage' 
    else: 
        heritage_type = 'Not Heritage'

    
    # Add marker to the map NOTE USE OF MARKER CLUSTER HERE
    folium.Marker(location=[lat, lon], 
                  tooltip=name,
                  icon=icon,
                  tags=[heritage_type]
    ).add_to(marker_cluster)


categories = ['Heritage','Not Heritage']
plugins.TagFilterButton(categories).add_to(m)

m
```

## Part 2: Open Street Map Data Gathering

Open Street Map is a global open-source map with lots of free tools for mapping. Here, we will use OSMnx to download and play with some data. 
Details of the type of information you can get from OSM can be found [here](https://wiki.openstreetmap.org/wiki/Map_features)
Documentation for the OSMnx package can be found [here](https://osmnx.readthedocs.io/en/stable/user-reference.html). 


### Getting features from a OpenStreetMap ID location

Using [Nominatim](https://nominatim.openstreetmap.org/ui/search.html), you can find the OSM ID number of many areas. Here, I searched Seattle Center, and found the matching ID number for the polygon I wanted. 

```python
# get the Seattle Center geometry
seattlecenter = ox.geocode_to_gdf('W4755066', by_osmid = True)

amenities = ox.features_from_polygon(seattlecenter["geometry"].iloc[0], tags={"amenity":['cafe']})

amenities = gpd.GeoDataFrame.from_features(amenities)
amenitypts = amenities[amenities.geom_type == 'Point']
amenitypts['lon']=amenitypts['geometry'].x
amenitypts['lat']=amenitypts['geometry'].y
amenitypts.head()
```

```python
alt.Chart(amenitypts).mark_circle(
    size = 10
).encode(
    latitude='lat:Q',
    longitude='lon:Q',
    color = 'amenity:N'
).project("mercator")
```

```python
# get the Seattle Center geometry
seattlecenter = ox.geocode_to_gdf('W4755066', by_osmid = True)

highways = ox.features_from_polygon(seattlecenter["geometry"].iloc[0], tags={"highway":True})
highways = highways[highways.geom_type != 'Point']

highways = gpd.GeoDataFrame.from_features(highways)
highways.head()
```

```python
alt.Chart(highways).mark_geoshape(
    fill='none',
    strokeWidth=2
).encode(
    stroke = 'highway:N'
).project('identity')
```

### Getting features from nearby a point

You can also choose a point and radius around which to get the features. Here I choose a point in downtown Seattle and a distance of 500 meters. 

```python
mypoint = [47.6038321,-122.330062] # center point in lat lon
mydist = 500 # distance in m

bikeparking = ox.features_from_point(center_point = mypoint, 
                                     tags={"amenity":['bicycle_parking']}, 
                                     dist = mydist)

bikeparking = gpd.GeoDataFrame.from_features(bikeparking)
bikeparking['lon']=bikeparking['geometry'].x
bikeparking['lat']=bikeparking['geometry'].y

apartments = ox.features_from_point(center_point = mypoint, 
                                     tags={"building": ["apartments"]}, 
                                     dist = mydist)
bikeplot = alt.Chart(bikeparking).mark_circle(
    stroke = 'red',
    fill = 'none',
    size = 4
).encode(
    latitude='lat:Q',
    longitude='lon:Q'
)

apartplot = alt.Chart(
    gpd.GeoDataFrame.from_features(apartments)
).mark_geoshape(fill='lightgray').project("identity")

alt.layer(bikeplot, apartplot).resolve_scale(color='independent').properties(width=800,height=800)
```

### Plotting Open Street Map Points on a Folium Plot

The following example shows one use case of plotting individual points downloaded from OSM onto an interactive Folium map.

```python
# get Seattle geometry
seattle = ox.geocode_to_gdf('R237385', by_osmid = True)

cafes = ox.features_from_polygon(seattle["geometry"].iloc[0], tags={"amenity":['cafe']})
cafes['geometry'] = cafes['geometry'].representative_point()
cafes = gpd.GeoDataFrame.from_features(cafes)

cafes['lon'] = cafes['geometry'].x
cafes['lat'] = cafes['geometry'].y

# cafes.shape
# print(cafes.columns.tolist())

cafes.head()
```

```python
m = folium.Map(location=[47.6, -122.3], 
               tiles='cartodbpositron',
               zoom_start=12, control_scale=True)

marker_cluster = plugins.MarkerCluster().add_to(m)

for idx, row in cafes.iterrows():
    # Get lat and lon of points
    lon = row['lon']
    lat = row['lat']
    
    # Add marker to the map NOTE USE OF MARKER CLUSTER HERE
    folium.Marker(
        location=[lat, lon],
        tooltip = row['name']
    ).add_to(marker_cluster)
m

```

### Plotting Open Street Map Polygons on a Folium Plot

The following example shows how to plot polygon geometry types downloaded from OSM.

```python
# get the Seattle geometry
seattle = ox.geocode_to_gdf('R237385', by_osmid = True)

# collect list of parks
cafes = ox.features_from_polygon(seattle["geometry"].iloc[0], tags={"leisure":['park']})
cafes = cafes[cafes.geom_type == 'Polygon'] # filter to only polygon geometires
cafes = gpd.GeoDataFrame.from_features(cafes) # convert to GeoDataFrame

# initialize map
m = folium.Map(location=[47.6, -122.3], 
               tiles='cartodbpositron',
               zoom_start=12, control_scale=True)

for idx, row in cafes.iterrows():

    # get coordinates of the polygon geometry into a plain list
    x, y = row['geometry'].exterior.coords.xy
    locs = tuple(zip(y,x))
    
    name = row['name']
    
    folium.Polygon(
        locations=locs,
        color="darkgreen",
        weight=1,
        fill_color="lightgreen",
        fill_opacity=0.5,
        fill=True,
        tooltip=name
    ).add_to(m)

m
```
