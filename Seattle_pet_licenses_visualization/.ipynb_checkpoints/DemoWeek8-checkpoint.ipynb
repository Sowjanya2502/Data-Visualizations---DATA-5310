# Working with Text Data
There are many ways to manipulate text in any programming language, and there are many packages for it in Python alone. Here we will be doing our work in Pandas to accommodate how we normally import and manipulate data. The Pandas documentation is extremely useful for the nuanced options and parameters for string manipulation!

Given a set of strings: 

    I love to ride my bike all day. 
    
    I love to play guitar. 
    
    I like eating. 
    
We may want to detect which ones say "bike". We may also want to replace that word with "bicyle" instead. 

---

First, we will make a Pandas series to hold our text. 

```python
se = pd.Series(['I love to ride my bike all day ','I love to play guitar','I like eating'])
se
```

Starting with our series, using `data.str.function()` will apply the string functions to the data, where `data` is the name of your variable.

--- 

## Cleaning Text Data

In order to get your data into a consistent format, it can be nice to know some of these functions. 

### Capitalizing and Lowercasing Words

`str.upper()` and `str.lower()` will easily capitalize/lowercase all your text. 

```python
se.str.upper()
```

```python
se.str.lower()
```

### Stripping Whitespaces

`str.strip()` will get rid of leading and trailing whitespaces for us. Notice the subtle difference when the final space is removed from the first entry. `lstrip` and `rstrip` also exist to separate leftward and rightward spaces. 

```python
se.str.strip()
```

### Splitting 

`str.split()` can split the string into multiple parts based on a token. You can also expand the string into multiple columns, for instance, if you wanted to separate first and last names. Or, you can access one element of the split with `str.get()`.

```python
se.str.split(' ')
```

```python
se.str.split(' ').str.get(2)
```

---
## Searching and Replacing

### Replacing

We can seek and replace entire sets of the string with `str.replace()`. 

```python
se.str.replace('bike','bicycle')
```

We can also chain together string commands in succession. Here, I strip the whitespaces and then replace like with love. 

**In another cell, in one line, capitalize all the letters and replace the word love with something else.**

```python
se.str.strip().str.replace('like','love')
```

### Contains

`str.contains()` will give boolean responses for whether that string *contains* the test string you input. By contains, we mean that the longer string has the test string somewhere in it, rather than the entire strings being identical.
**Add another cell below to find which strings contain the phrase 'I love'.**

```python
se.str.contains('eating')
```

### Match and Extract
`str.match()` also gives a boolean response, but now tests if the whole strings match. Notice that 'eating' does not match any entire string, converse to `str.contains()` above. 

`str.extract()` takes the boolean match and instead returns the matched string. Here, we have to inlcude parenthesis to indicate the text we want to capture. This will be more useful with regular expressions below. 

```python
se.str.match('I like eating')
```

```python
se.str.match('eating')
```

```python
se.str.extract('I like (eating)')
```

### Find and Find All
`str.find()` will return the index location of the **first** match of a string or pattern. Here, searching for 'a' will give the first location in the string where the letter 'a' appears.

`str.findall()` will find all matches and return the matches only. This will be useful when searching for patterns rather than exact strings. 

```python
se.str.find('a')
```

```python
se.str.findall('a')
```

---
## Regular Expressions 

Regular expressions are a way of generalizing text to do flexible searches with defined patterns. Say we are looking for prices in restaurant reviews. A regular expression would let us search for digit values within the text. 

There are infinitely many possibilities for searching with regular expressions, I suggest going to a regex tester like [Regex101](https://regex101.com/) to test out options. 


```python
food = pd.Series(["I can't believe I paid $12.56 for a bottle of water",
                  "The pasta was worth every one of the 10 dollars we paid.",
                 "14534545 minute wait for the table."])
food
```

The regular expression for a digit in Python is `\d`. In this example, I will search for the presence of two digits in a row. 

```python
food.str.contains('\d\d')
```

Since all the examples have cases where two digits appear in succession, we get True for all rows. If we look for two digits preceded by a dollar sign, we see that only the first value is True. Note that we can't just use a dollar sign, `$`, we have to use `\$` to escape the meaning that regex uses for `$`, which indicates the beginning of the string. 

```python
food.str.contains('\$\d\d')
```

You can extract the pattern as well with `str.extract()`. Here, you must include parenthesis to "capture" the text. Use `str.extractall()` if you want all the matches rather than the first one only. Notice that in the zeroeth index, we get two matches, the first two digits, and the second two digits in the price. 

```python
food.str.extract('(\d{2})')
```

```python
food.str.extractall('(\d{2})')
```

The use of the capture group shows up more clearly when we specify further patterns. Here, we look for 1 to 4 digits with a dollar sign out front, returning only the digits. This can be useful if we wanted to then convert the column to a numeric data type, where the dollar sign would not be recognized.  

```python
food.str.extract('\$(\d{1,4})')
```

## Fuzzy Matching 

Now, we will look at some less strict string matches. At times, knowing which exact words or phrases appear in text is nearly impossible. Knowing how to find a "close enough" match is extremely useful. 

Getting the ratio of a string to a test string is comparing their total percent match, including punctuation. The number returned is the percentage that the two strings match, with 100 being a perfect match. 

```python
fuzz.ratio("this is a test", "this is a test!")
```

The partial ratio allows the string to be compared to a *subset* of the test string. Try changing the text outside the test string to see how it affects the match. 

```python
fuzz.partial_ratio("this is a test", "this is a test!")
```

The token sort ratio first sorts the set of words, then compares them directly. In this case, the order does not matter. 

```python
fuzz.token_sort_ratio("fuzzy wuzzy was a bear", "wuzzy fuzzy was a bear")
```

The token set ratio takes the set of unique words into account, regardless of repetition.   

```python
fuzz.token_set_ratio("fuzzy was a bear", "fuzzy fuzzy was a bear")
```

The process function gets the closest options out of a list of possibilities based on their distances. 

```python
choices = ["Atlanta Falcons", "New York Jets", "New York Giants", "Dallas Cowboys"]
process.extract("new york jets", choices, limit=1)
```

```python
process.extractOne("cowboys", choices)
```

## Plotting Text Data

There are many packages that we could use to make word clouds, today we will use `wordcloud` for its simplicity. Here are a few examples of what we can do with this package, with code based on their examples and documentation.
There are of course, [detailed documentation and more examples](https://amueller.github.io/word_cloud/index.html). 

### Frankenstein Data

Because of US copyright laws, many older classical literature works are now freely available online. [Project Gutenberg](https://www.gutenberg.org/) is one excellent source of these texts. Today, we will use the example of Franeknstein by Mary Shelley. It is regarded as the birth of the science fiction genre and was first published in 1818. 

Below, we use a web scraper and HTML parser to download this simple online text format novel. We then parse it into a string and clean up some mismatched characters. 

```python
frank = 'https://www.gutenberg.org/files/84/84-h/84-h.htm'
html = requests.get(frank).text
soup = BeautifulSoup(html,'html.parser') 
txt = ' '.join([item.text for item in soup.findAll("p")])
to_remove = {
    '\r\n':' ',
    '\n': '',
    'â\x80\x94': '--',
    'â\x80\x99': "'",
    'â\x80\x9c':'"',
    'â\x80\x9d':'"'
}
for char in to_remove.keys():
     txt = txt.replace(char, to_remove[char])
print(txt[:500])  
```

### Word Cloud Generation

We can generate a word cloud based on the count of words in the document. Because many words may appear more commonly but hold little meaning, like "the" and "a", these words are automatically excluded from the cloud. You can add words to this list of exclusions to suit the needs of your corpus of text. 

```python
wordcloud = WordCloud().generate(txt)
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis("off")
plt.show()
```

### Using a mask to shape the word cloud

We can plot our word cloud not just in a rectangle but also in any shape we want. We could generate a mask using numbers from a generated matrix or from an image. Here we download a sample image of Frankenstein's head to use as our mask.

```python
img_data = requests.get('https://cdn-icons-png.flaticon.com/512/597/597434.png').content
with open('frankenstein.png', 'wb') as handler:
    handler.write(img_data)

mask = np.array(Image.open("frankenstein.png"))
plt.imshow(mask)
plt.axis("off")
plt.show()
```

Then, we can generate our word clous using this mask. We add a margin so the words aren't squished to the edges.

```python
wc = WordCloud(mask=mask, margin=10, random_state=1).generate(txt)
plt.imshow(wc)
plt.axis("off")
plt.show()
```

### Using custom colors

We can customize our color scheme using the optional parameter `colormap`, with the options given by the [matplotlib color schemes](https://matplotlib.org/stable/users/explain/colors/colormaps.html). Alternatively, we can define a custom color function, here creating a random HSL-spec color scheme of grays. We may want to do a custom scheme to avoid the random selection of colors at the edges of the color spectrum. For example, if we used the parameter `colormap = 'Greys'` here, it may plot words in black, which would not be very visible against the dark background.

```python
def grey_color_func(word, font_size, position, orientation, random_state=None, **kwargs):
    return "hsl(0, 0%%, %d%%)" % np.random.randint(60, 100)

wcrecolor = wc.recolor(color_func=grey_color_func).to_image()
plt.imshow(wcrecolor)
plt.axis("off")
plt.show()
```
